```python
import cv2
import matplotlib.pyplot as plt

# Load image - CHANGE THIS PATH to your desired image file
img = cv2.imread('/content/block 2.jpg')

# Check if the image was loaded successfully
if img is None:
    print("Error: Could not load the image. Please check the file path.")
else:
    # Convert BGR → RGB for matplotlib display
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()
```

```python
if img is not None:
    # Convert to Grayscale , SIFT works on grayscale intensity.
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.imshow(gray, cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()
else:
    print("Error: The image 'img' was not loaded in a previous step, cannot convert to grayscale.")
```

```python
# Create the SIFT Detector
sift = cv2.SIFT_create()
```

```python
# Detect Keypoints and Descriptors
if 'gray' in locals() and gray is not None:
    keypoints, descriptors = sift.detectAndCompute(gray, None)

    print("Number of keypoints:", len(keypoints))
    print("Descriptor shape:", descriptors.shape)
else:
    print("Error: 'gray' image is not defined. Please ensure the image was loaded and converted to grayscale successfully in previous steps.")
```

```python
# Draw Keypoints
if 'img_rgb' in locals() and img_rgb is not None and 'keypoints' in locals() and keypoints is not None:
    img_keypoints = cv2.drawKeypoints(
        img_rgb,
        keypoints,
        None,
        flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
    )

    plt.figure(figsize=(12,8))
    plt.imshow(img_keypoints)
    plt.title("SIFT Keypoints")
    plt.axis("off")
    plt.show()
else:
    print("Error: 'img_rgb' or 'keypoints' not defined. Please ensure the image was loaded and keypoints were detected successfully in previous steps.")
```

### Feature Matching between the two images

```python
import cv2
import matplotlib.pyplot as plt

# Create a Brute-Force Matcher object
# cv2.NORM_L2 is used for SIFT descriptors
bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=False)

# Match descriptors. knnMatch returns k best matches for each descriptor
matches = bf.knnMatch(descriptors, descriptors2, k=2)

# Apply ratio test (Lowe's ratio test)
good_matches = []
for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

print(f"Found {len(good_matches)} good matches after ratio test.")
```

```python
# Draw the matches
# The images need to be combined horizontally or vertically for visualization
# We'll use the RGB versions of the images (img_rgb and img2_rgb)

# Ensure both images are RGB before drawing matches
# cv2.drawMatchesKnn expects images in the same format. It's safer to ensure they are BGR if using cv2 for display
# However, since we are using matplotlib, RGB is preferred.

# If you run into issues, convert back to BGR for cv2.drawMatchesKnn
# img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
# img2_bgr = cv2.cvtColor(img2_rgb, cv2.COLOR_RGB2BGR)

matched_img = cv2.drawMatches(
    img_rgb,        # First image (in RGB format for matplotlib)
    keypoints,      # Keypoints from the first image
    img2_rgb,       # Second image (in RGB format for matplotlib)
    keypoints2,     # Keypoints from the second image
    good_matches,   # Good matches found
    None,           # Output image (automatically created if None)
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

plt.figure(figsize=(20, 10)) # Adjust figure size for better visualization
plt.imshow(matched_img)
plt.title('SIFT Feature Matches (Ratio Test)')
plt.axis('off')
plt.show()
```

### Canny Edge Detection for Visual Comparison

To explicitly visualize edges in both images, we can use the Canny edge detection algorithm. This will allow us to see how the SIFT keypoints and their matches relate to the prominent contours and lines in each image.

```python
import cv2
import matplotlib.pyplot as plt

# Apply Canny edge detection to the grayscale images
# You might need to adjust the threshold values (100, 200) based on your images
edges1 = cv2.Canny(gray, 100, 200)
edges2 = cv2.Canny(gray2, 100, 200)

plt.figure(figsize=(20, 10))

plt.subplot(121)
plt.imshow(edges1, cmap='gray')
plt.title('Image 1: Canny Edges')
plt.axis('off')

plt.subplot(122)
plt.imshow(edges2, cmap='gray')
plt.title('Image 2: Canny Edges')
plt.axis('off')

plt.show()
```

Now that we have both the SIFT feature matches and the Canny edge maps, you can visually compare them. Notice how many of the SIFT keypoints (and thus the matched lines) often originate from or lie along the strong edges detected by the Canny algorithm. This shows that SIFT features often correspond to these critical structural elements in an image.

While this visualization helps in *identifying* where matched keypoints fall in relation to edges, directly "matching edges" in a more abstract sense (e.g., matching entire edge segments or contours) is a more advanced task that goes beyond typical SIFT applications and would require different techniques like contour matching or geometric transformations.
